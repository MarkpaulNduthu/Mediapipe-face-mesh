```python
import mediapipe as mp
import cv2
```

```python
mp_handpose_solution = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils
```

INITIALIZE THE CAMERA

```python
cap = cv2.VideoCapture(0)
```

```python
try:
    with mp_handpose_solution.Hands(
    max_num_hands=2,
    min_detection_confidence=0.2,
    min_tracking_confidence=0.2
    ) as hands:
        while cap.isOpened():
            success, frame = cap.read()
            if not success:
                print("Something went wrong!")
                break
            rgb_frame = cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
            result = hands.process(rgb_frame)
            if result.multi_hand_landmarks:
                for landmark in result.multi_hand_landmarks:
                    mp_drawing.draw_landmarks(
                        image=frame,
                        landmark_list=landmark,
                        connections=mp_handpose_solution.HAND_CONNECTIONS,
                        landmark_drawing_spec=mp_drawing.DrawingSpec(color=(0,255,0),thickness=2,circle_radius=2),
                        connection_drawing_spec=mp_drawing.DrawingSpec(color=(0,0,255),thickness=2,circle_radius=2)
                    )
            bgr_frame = cv2.cvtColor(frame, cv2.COLOR_RGB2BGR)
            cv2.imshow("Hand pose detection", bgr_frame)
            if cv2.waitKey(1) & 0XFF == ord('q'):
                break
finally:
    cap.release()
    cv2.destroyAllWindows()
        
```
